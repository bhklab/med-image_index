```python
description = """
The PANORAMA study data set consists of ~3000 cases from three dutch centers (Radboud University Medical Center, University Medical Center Groningen, Ziekenhuis Groep Twente), one center in Sweden (Karolinska Institutet), and one center in Norway (Haukeland University Hospital). Institutional review boards of all centers have waived the need for informed patient consent, with respect to the retrospective scientific use of anonymized clinical data in this challenge. All images are contrast-enhanced CT (CECT) scans in the portal-venous phase, as this is the most commonly acquired phase in routine abdominal imaging. All exams are from patients undergoing CECT without a history of pancreatic cancer treatment, and without any prior positive PDAC histopathology findings. All cases also contain the clinical parameters: age, sex, study date, and scanner. Missing values are possible if they were not recorded in the original dicom headers.
"""
```

```python
import zipfile
from glob import glob
import os

def extract_zip(zip_path, extract_to=None):
    """
    Extract a zip file to a specified directory
    
    Args:
        zip_path: Path to the zip file
        extract_to: Directory to extract to (default: same directory as zip)
    """
    if extract_to is None:
        extract_to = os.path.dirname(zip_path)
    
    # Create extraction directory if it doesn't exist
    os.makedirs(extract_to, exist_ok=True)
    
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
        print(f"Extracted {zip_path} to {extract_to}")

for zip_file in glob(f"temp_data/{DATASET_NAME}/*.zip"):
    print(zip_file)
    # Get the base name without extension to create a folder with the same name
    zip_name = os.path.splitext(os.path.basename(zip_file))[0]
    extract_to = os.path.join(f"temp_data/{DATASET_NAME}", zip_name)
    extract_zip(zip_file, extract_to)
```

```python


# We have two different scan and mask patterns that i need to capture. We are going to do 2 crawls and combine the indexes.

crawler = Crawler(
    nifti_dir=temp_data_path,
    mask_name_pattern="{label_type}_labels/{reference_id}.nii.gz",
    scan_name_pattern="batch_{batch_num}/{reference_id}_0000.nii.gz",
    output_dir=indexed_datasets.path,
    n_jobs=10,
    force=True,
    deep=False,
)
crawler.crawl()

index_df = pd.read_csv(indexed_datasets.path / DATASET_NAME / "index.csv")

crawler = Crawler(
    nifti_dir=temp_data_path,
    mask_name_pattern="{label_type}_labels/{reference_id}.nii.gz",
    scan_name_pattern="batch_{batch_num}/batch_{batch_num2}/{reference_id}_0000.nii.gz",
    output_dir=indexed_datasets.path,
    n_jobs=10,
    force=True,
    deep=False,
)
crawler.crawl()

index_df2 = pd.read_csv(indexed_datasets.path / DATASET_NAME / "index.csv")

index_df = pd.concat([index_df, index_df2], axis=0, ignore_index=True)

```

```python
#index_df = pd.read_csv(indexed_datasets.path / DATASET_NAME / "index.csv")
print(len(index_df))
index_df = index_df.drop_duplicates(subset=['label_type', 'reference_id', 'file_type'])
print(len(index_df))
```

```python
print(len(index_df))
index_df
```

```python
def map_modality(row):
    if row["file_type"] == "scan":
            return "CT"
    else:
        return "SEG"
    
def map_roi(row):
    if row["file_type"] == "scan":
            return ""
    else:
        return "PDAC Lesion,Veins,Arteries,Pancreas parenchyma,Pancreatic duct,Common bile duct"
    
index_df["Modality"] = index_df.apply(map_modality, axis=1)
index_df["ROINames"] = index_df.apply(map_modality, axis=1)

index_df["BodyPartExamined"] = "ABDOMEN"


```

```python
clinical_df = pd.read_excel(temp_data_path / "clinical_information.xlsx")

combined_df = index_df.merge(clinical_df, left_on="reference_id", right_on="PANORAMA_study_id", how="left")
```

```python
combined_df = combined_df.drop(columns=["batch_num", "batch_num2"])
```

```python
from imgnet.collections.store import IndexedDatasets
from imgindex.model import validate_index
from imgindex.parquet import csv_to_parquet

index_df = combined_df
index_df.rename(columns={"reference_id": "SampleID"}, inplace=True)
index_df["SampleID"] = index_df["SampleID"].astype(str)
index_df.to_csv(indexed_datasets.path / DATASET_NAME / "index.csv", index=False)
csv_to_parquet(index_df, indexed_datasets.path / DATASET_NAME)
validate_index(index_df, "nifti", lazy=True)
```
